# Popularity Based system

```python
# top 50 books (conditions : books have minimum 250 rating)
```

```python
ratings_with_name = ratings.merge(books, on='ISBN')
```

```python
ratings_with_name.info()
```

```python
ratings_with_name.shape
```

```python
num_rating_df = ratings_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating' : 'Num_Ratings'},inplace=True)
num_rating_df
```

```python
avg_rating_df = ratings_with_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
avg_rating_df.rename(columns={'Book-Rating' : 'Avg_Ratings'},inplace=True)
avg_rating_df
```

```python
popular_df = num_rating_df.merge(avg_rating_df, on="Book-Title")
popular_df
```

```python
popular_df = popular_df[popular_df['Num_Ratings']>=250].sort_values("Avg_Ratings", ascending=False).head(50)
```

```python
popular_df
```

```python
popular_df = popular_df.merge(books, on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Image-URL-M','Num_Ratings','Avg_Ratings']]
```

```python
popular_df
```

# Collaborative Filtering Based Recommender System

```python
x = ratings_with_name.groupby('User-ID').count()['Book-Rating'] > 200
frequent_users = x[x].index
```

```python
filtered_rating = ratings_with_name[ ratings_with_name['User-ID'].isin(frequent_users)]
```

```python
y = filtered_rating.groupby('Book-Title').count()['Book-Rating'] >= 50
famous_books = y[y].index
```

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_books)]
```

```python
pt = final_ratings.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
```

```python
pt
```

```python
pt.fillna(0,inplace=True)
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_score = cosine_similarity(pt)
```

```python
similarity_score.shape
```

```python
def recommend(book_name):
  # index fetch
  index = np.where(pt.index==book_name)[0][0]
  similer_items = sorted(list(enumerate(similarity_score[index])),key = lambda x : x[1],reverse=True)[1:6] # first not select because , give best score with self cosine similarity

  data = []
  for i in similer_items:
    item = []
    temp_df = books[books['Book-Title'] == pt.index[i[0]]]
    item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Title'].values))
    item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Author'].values))
    item.extend(list(temp_df.drop_duplicates('Book-Title')['Image-URL-M'].values))

    data.append(item)
  return data
```

```python
recommend("Harry Potter and the Prisoner of Azkaban (Book 3)")
```
